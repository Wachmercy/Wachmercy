```python

#  NLP with spaCy – Amazon Product Reviews




import spacy
from spacy.matcher import Matcher

# Load the English NLP model
nlp = spacy.load("en_core_web_sm")


# Step 2: Sample Amazon Review Data

reviews = [
    "I love my new Samsung Galaxy phone! The camera quality is excellent.",
    "This Apple MacBook is overpriced but performs really well.",
    "Sony headphones are amazing, especially for the price!",
    "I’m disappointed with my Dell laptop, battery drains too fast.",
]


# Step 3: Named Entity Recognition (NER)

print(" NAMED ENTITY RECOGNITION \n")
for review in reviews:
    doc = nlp(review)
    print(f"🔹 Review: {review}")
    print("Extracted Entities:")
    for ent in doc.ents:
        print(f" - {ent.text} ({ent.label_})")
    print("-" * 60)


# Step 4: Rule-Based Sentiment Analysis

print("\n RULE-BASED SENTIMENT ANALYSIS \n")

positive_words = ["love", "excellent", "amazing", "good", "great", "well"]
negative_words = ["bad", "poor", "overpriced", "disappointed", "slow", "drains"]

def rule_based_sentiment(text):
    """Simple rule-based sentiment scoring"""
    text_lower = text.lower()
    pos_score = sum(word in text_lower for word in positive_words)
    neg_score = sum(word in text_lower for word in negative_words)
    if pos_score > neg_score:
        return "Positive 😊"
    elif neg_score > pos_score:
        return "Negative 😞"
    else:
        return "Neutral 😐"

# Apply sentiment function
for review in reviews:
    sentiment = rule_based_sentiment(review)
    print(f"Review: {review}\n→ Sentiment: {sentiment}\n")


# Step 5: Ethical Considerations (Documentation)
# ------------------------------------------------------------
"""
================== ETHICAL CONSIDERATIONS ==================

⚠️ Potential Biases:
1. Dataset Bias:
   - Reviews may reflect cultural tone or sarcasm (e.g., “not bad” can be positive).
   - Reviews in mixed languages may confuse the model.

2. Model Bias:
   - Pretrained spaCy models (en_core_web_sm) trained mostly on Western text corpora.
   - Non-English brand names or products may be missed or mislabeled.

3. Sentiment Rule Bias:
   - Simple positive/negative word lists ignore context (e.g., “too good to be true” is negative).

🧰 Mitigation Tools:
- TensorFlow Fairness Indicators:
  → Helps visualize and compare model performance across demographic or product subgroups.

- spaCy Rule-Based Systems:
  → Allow adding custom entity-matching rules for specific products/brands.

Example Custom Rule:
------------------------------------------------------------
matcher = Matcher(nlp.vocab)
pattern = [{"TEXT": {"REGEX": "Galaxy|MacBook|headphones|laptop"}}]
matcher.add("PRODUCT", [pattern])
------------------------------------------------------------
This extends detection to domain-specific terms, reducing missed entities.

============================================================
"""
```
